# Brique 1 — Exercices faciles

**Comment utiliser ce notebook.** Chaque exercice a deux cellules : une où tu écris ta solution (elle contient
`raise NotImplementedError`, une instruction qui arrête volontairement l'exécution avec une erreur — c'est un
espace réservé qui signale "il manque du code ici", à remplacer entièrement par ta réponse), puis une cellule de
test juste après.

**Ce que fait une cellule de test** : le mot-clé `assert` vérifie qu'une condition est vraie ; si elle est
fausse, Python arrête tout avec une erreur `AssertionError`. Donc si tu lances la cellule de test et qu'elle
affiche `OK` sans erreur, ta fonction a réussi tous les cas vérifiés. C'est le même principe que les `grader.py`
de tes TD `Data analysis` : tu écris ta solution, un programme séparé vérifie si elle est correcte, sans que tu
aies à comparer les résultats toi-même à l'œil.
