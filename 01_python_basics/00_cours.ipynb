# Brique 1 — Rappel Python de base

Tu connais déjà la programmation (Java). Ce notebook n'explique pas "c'est quoi une boucle" — il fixe le

## 1. Types de base

Pas de déclaration de type (vu en Brique 0 : typage dynamique). `None` remplace `null`.

```python
entier = 42
flottant = 3.14
texte = "bonjour"
booleen = True
rien = None

for valeur in (entier, flottant, texte, booleen, rien):
    print(valeur, "->", type(valeur))
```

Une différence utile avec Java : la division `/` renvoie toujours un flottant, même entre deux entiers.
`//` est la division entière, `%` le modulo, `**` la puissance (pas de `Math.pow` à importer).

```python
print(7 / 2)    # 3.5  (en Java, 7 / 2 donnerait 3 si les deux sont des int)
print(7 // 2)   # 3    division entière explicite
print(7 % 2)    # 1
print(2 ** 10)  # 1024
```

## 2. Structures de données — laquelle choisir

| Structure | Équivalent Java | Ordonnée ? | Modifiable ? | Doublons ? | Quand l'utiliser |
|---|---|---|---|---|---|
| `list` | `ArrayList` | Oui | Oui | Oui | La collection par défaut, usage général |
| `tuple` | — (pas d'équivalent direct) | Oui | **Non** | Oui | Un groupe fixe de valeurs (coordonnées, retour multiple d'une fonction) |
| `dict` | `HashMap` | Oui (depuis Python 3.7, ordre d'insertion) | Oui | Clés uniques | Association clé → valeur |
| `set` | `HashSet` | Non | Oui | **Non** | Appartenance rapide, éliminer les doublons |

```python
ma_liste = [1, 2, 3]
ma_liste.append(4)

mon_tuple = (48.8566, 2.3522)  # latitude, longitude — un "paquet" de valeurs qui ne doit jamais changer

mon_dict = {"nom": "Youssef", "annee": "3A"}
mon_dict["ecole"] = "Polytechnique"  # ajoute une clé

mon_set = {1, 2, 2, 3, 3, 3}  # les doublons disparaissent automatiquement
print(ma_liste, mon_tuple, mon_dict, mon_set)
```

**Indexation et slicing**, propre à Python (pas d'équivalent aussi direct en Java pour les tableaux) :

```python
lettres = ["a", "b", "c", "d", "e"]
print(lettres[0])    # premier élément
print(lettres[-1])   # dernier élément (l'index négatif compte depuis la fin)
print(lettres[1:3])  # slicing : du 2e (inclus) au 4e (exclu)
print(lettres[::-1]) # astuce classique : inverser la liste
```

## 3. Contrôle de flux et comprehensions

L'indentation remplace les accolades `{ }` — ce n'est pas du style, c'est **la syntaxe elle-même**.

```python
note = 14
if note >= 16:
    mention = "très bien"
elif note >= 14:
    mention = "bien"
else:
    mention = "passable"
print(mention)

for i in range(5):        # équivalent de for (int i = 0; i < 5; i++)
    print(i, end=" ")
print()

for lettre in "abc":       # on peut itérer directement sur beaucoup de types (chaînes, listes, dicts...)
    print(lettre, end=" ")
print()
```

**List comprehension** : une façon très idiomatique en Python de construire une liste à partir d'une boucle,
en une seule ligne lisible. Tu la croiseras partout dans du code Python professionnel.

```python
carres = [n ** 2 for n in range(10)]
print(carres)

carres_pairs = [n ** 2 for n in range(10) if n % 2 == 0]  # avec un filtre
print(carres_pairs)

# Équivalent explicite avec une boucle classique, pour voir que c'est la même chose :
carres_pairs_v2 = []
for n in range(10):
    if n % 2 == 0:
        carres_pairs_v2.append(n ** 2)
print(carres_pairs_v2)
```

## 4. Fonctions

Pas de surcharge de méthodes (comme en Java) — à la place, des **valeurs par défaut** et des **arguments par
mot-clé**.

```python
def saluer(nom, langue="fr"):   # langue a une valeur par défaut
    if langue == "fr":
        return f"Bonjour {nom}"
    return f"Hello {nom}"

print(saluer("Youssef"))                 # utilise la valeur par défaut
print(saluer("Youssef", langue="en"))    # argument par mot-clé, plus lisible qu'un argument positionnel
```

`*args` capture un nombre variable d'arguments positionnels, `**kwargs` un nombre variable d'arguments par
mot-clé. Tu les verras énormément dans les signatures de bibliothèques (scikit-learn, pandas) en Brique 4.

```python
def somme(*nombres):
    return sum(nombres)

print(somme(1, 2, 3, 4))  # autant d'arguments que tu veux

def decrire(**infos):
    for cle, valeur in infos.items():
        print(f"{cle} : {valeur}")

decrire(nom="Youssef", annee="3A", ecole="Polytechnique")
```

`lambda` : une fonction anonyme d'une seule expression, utile comme argument court (par exemple pour trier).

```python
personnes = [("Alice", 30), ("Bob", 25), ("Chloé", 35)]
personnes_triees = sorted(personnes, key=lambda p: p[1])  # trie par âge (2e élément du tuple)
print(personnes_triees)
```

## 5. Chaînes de caractères

Les **f-strings** (`f"..."`) sont la façon moderne de construire une chaîne avec des valeurs insérées —
équivalent de la concaténation `+` ou de `String.format` en Java, mais plus lisible.

```python
nom = "Youssef"
age = 22
print(f"{nom} a {age} ans")
print(f"Dans 5 ans, {nom} aura {age + 5} ans")  # on peut mettre une expression directement dans les accolades

phrase = "  Bonjour le Monde  "
print(phrase.strip())        # retire les espaces en début/fin
print(phrase.strip().lower())
print(phrase.strip().split(" "))  # découpe en liste de mots
```

## 6. Gestion des erreurs

Équivalent du `try/catch` Java, avec une syntaxe légèrement différente.

```python
def diviser(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        print("Division par zéro !")
        return None

print(diviser(10, 2))
print(diviser(10, 0))
```

## 7. Fichiers

Le mot-clé `with` garantit que le fichier est fermé automatiquement, même en cas d'erreur — équivalent de
try-with-resources en Java.

```python
with open("exemple.txt", "w", encoding="utf-8") as f:
    f.write("Bonjour depuis Python\n")

with open("exemple.txt", "r", encoding="utf-8") as f:
    contenu = f.read()
print(contenu)
```
