# Brique 1 — Rappel Python de base, construit depuis zéro

Tu sais déjà programmer (Java), donc les *idées* derrière ce qu'on va voir (une boucle, une fonction) ne sont pas
nouvelles pour toi. Ce qui change, c'est **la syntaxe précise de Python**, et quelques façons de penser propres à
ce langage. On construit chaque brique de vocabulaire avant de s'en servir, sans rien supposer acquis.

Plan : valeurs et types → variables → opérateurs → structures de données → indexation → contrôle de flux →
fonctions → chaînes de caractères → erreurs → fichiers.

## 1. Valeurs et types de base

**Qu'est-ce qu'une "valeur" ?** C'est la chose la plus simple qui existe en programmation : une donnée concrète
que le programme manipule — le nombre `42`, le texte `"bonjour"`, le fait que quelque chose soit vrai ou faux.

**Et un "type" ?** On l'a déjà construit en Brique 0 (section 7) : c'est une catégorie qui dit quelle sorte de
valeur on a, et donc ce qu'on a le droit de faire avec. Listons les types les plus basiques de Python, chacun
avec sa définition en langage simple :

- `int` (entier) : un nombre sans virgule, comme `42` ou `-7`.
- `float` (flottant) : un nombre à virgule, comme `3.14`.
- `str` (chaîne de caractères, de l'anglais *string*) : du texte, toujours entouré de guillemets, comme
  `"bonjour"`.
- `bool` (booléen) : seulement deux valeurs possibles, `True` (vrai) ou `False` (faux).
- `None` : une valeur spéciale qui représente "rien", "l'absence de valeur" — c'est l'équivalent direct de
  `null` en Java.

Vérifions ça avec la fonction `type()`, qui renvoie le type réel d'une valeur :

```python
print(type(42))
print(type(3.14))
print(type("bonjour"))
print(type(True))
print(type(None))
```

## 2. Les variables : juste un nom qui pointe vers une valeur

On a construit cette idée en détail en Brique 0 (section 7) : une variable Python n'est pas une boîte qui
contient une valeur avec un type fixé pour toujours (l'image qu'on a en tête venant de Java). C'est plutôt un nom
autocollant qu'on colle sur une valeur, et qu'on peut recoller plus tard sur une valeur complètement différente,
y compris d'un autre type. On crée une variable avec `=` (le signe égal) :

```python
age = 22          # 'age' pointe maintenant vers l'entier 22
print(age)

age = age + 1     # on calcule 22 + 1 = 23, puis on recolle 'age' sur cette nouvelle valeur
print(age)
```

Contrairement à Java, pas besoin d'écrire `int age = 22;` : tu n'annonces jamais le type à l'avance, Python le
déduit automatiquement à partir de la valeur que tu lui donnes.

## 3. Les opérateurs arithmétiques de base

Un **opérateur**, c'est juste un symbole qui désigne une opération à effectuer sur des valeurs (`+` pour
additionner, par exemple). La plupart te sont déjà familiers depuis Java, mais deux d'entre eux se comportent
différemment, donc on les construit avec attention :

```python
print(7 + 2)   # addition : 9
print(7 - 2)   # soustraction : 5
print(7 * 2)   # multiplication : 14
print(7 ** 2)  # puissance : 7 au carré, donc 49 (pas besoin d'importer Math.pow comme en Java)

# Attention ici, c'est la vraie différence avec Java :
print(7 / 2)   # division : 3.5 — TOUJOURS un résultat à virgule, même si les deux nombres sont des entiers
print(7 // 2)  # division entière : 3 — le symbole // dit explicitement "je veux seulement la partie entière"
print(7 % 2)   # modulo : 1 — le reste de la division entière de 7 par 2
```

En Java, `7 / 2` entre deux `int` te donnerait directement `3` (la partie entière, sans que tu l'aies demandé
explicitement). En Python, `/` donne **toujours** un résultat à virgule : si tu veux la division entière, tu dois
le dire explicitement avec `//`. C'est un piège classique quand on vient de Java.

## 4. Les structures de données : comment ranger plusieurs valeurs ensemble

Jusqu'ici, on n'a manipulé qu'une seule valeur à la fois. Une **structure de données**, c'est simplement une
façon de regrouper plusieurs valeurs ensemble, sous un seul nom. Python en propose quatre principales, et
chacune répond à un besoin différent. On les construit une par une.

### 4.1 La liste (`list`) — une collection ordonnée et modifiable

C'est la structure la plus simple et la plus utilisée : une suite de valeurs, dans un ordre précis, qu'on peut
modifier après sa création (ajouter, retirer, changer un élément). C'est l'équivalent le plus proche de
`ArrayList` en Java.

```python
fruits = ["pomme", "banane", "cerise"]
print(fruits)

fruits.append("datte")   # ajoute un élément à la fin
print(fruits)

fruits[0] = "abricot"    # remplace le premier élément
print(fruits)
```

### 4.2 Le tuple — une collection ordonnée, mais figée pour toujours

Un `tuple` ressemble beaucoup à une liste (un ordre de valeurs), mais avec une différence fondamentale : une
fois créé, **on ne peut plus jamais le modifier**. On appelle ça être **immuable** (en anglais, *immutable* —
retiens ce mot, il revient souvent). Une `list`, à l'inverse, est **muable** (*mutable*) : elle peut changer
après sa création, comme on vient de le voir avec `.append()`.

On écrit un tuple avec des parenthèses au lieu de crochets :

```python
coordonnees = (48.8566, 2.3522)  # latitude, longitude de Paris
print(coordonnees)

# coordonnees[0] = 0.0  # si tu décommentes cette ligne et la lances, Python refuse : TypeError
#                        # c'est exactement ça, être immuable : aucune modification possible, jamais
```

**Pourquoi utiliser un tuple plutôt qu'une liste, alors ?** Parce que l'immuabilité est parfois exactement ce
qu'on veut : ces deux coordonnées *forment* une position géographique précise, ensemble — ça n'aurait aucun sens
de changer seulement la latitude sans toucher à la longitude. Le tuple communique cette intention : "ces valeurs
vont ensemble, en bloc, et ne doivent jamais être modifiées séparément".

### 4.3 Le dictionnaire (`dict`) — une association clé → valeur

Imagine un vrai dictionnaire de langue : tu ne le lis pas page par page dans l'ordre pour trouver un mot, tu
cherches directement le mot qui t'intéresse (la **clé**) pour obtenir sa définition (la **valeur**). Un `dict`
Python fonctionne exactement comme ça : une collection de paires clé/valeur, où on accède directement à une
valeur à partir de sa clé, sans avoir à parcourir tout le reste. C'est l'équivalent de `HashMap` en Java.

```python
personne = {"nom": "Youssef", "annee": "3A"}
print(personne["nom"])          # accès direct à la valeur associée à la clé "nom"

personne["ecole"] = "Polytechnique"  # ajoute une nouvelle paire clé/valeur
print(personne)
```

### 4.4 Le set — une collection où chaque valeur n'existe qu'une seule fois

Imagine une liste d'invités à une fête : si tu ajoutes deux fois le même nom, ça ne fait toujours qu'un seul
invité réel. Un `set` fonctionne comme ça : une collection où **les doublons disparaissent automatiquement**, et
où (contrairement à la liste) il n'y a pas de notion d'ordre ni de position. On l'utilise surtout pour deux
choses : éliminer des doublons, ou vérifier très rapidement si une valeur est présente.

```python
invites = {"Alice", "Bob", "Alice", "Chloé", "Bob"}
print(invites)   # "Alice" et "Bob" n'apparaissent qu'une seule fois chacun, malgré les doublons écrits
```

## 5. Indexation et slicing — accéder à un ou plusieurs éléments précis

**L'indexation** : chaque élément d'une liste (ou d'un texte) a une position numérotée, en commençant à **0**
pour le premier élément (pas 1 — c'est une source d'erreur fréquente quand on commence).

```python
lettres = ["a", "b", "c", "d", "e"]
print(lettres[0])   # le premier élément : "a"
print(lettres[2])   # le troisième élément (position 2, en comptant depuis 0) : "c"
print(lettres[-1])  # un index négatif compte depuis la FIN : -1 est le dernier élément, "e"
```

**Le slicing** ("tranchage") : au lieu de récupérer un seul élément, on peut en récupérer plusieurs d'un coup,
avec la notation `debut:fin`. Règle importante à bien fixer : la position de `fin` n'est **jamais incluse**.

```python
print(lettres[1:3])   # positions 1 et 2 seulement (PAS la position 3) : ["b", "c"]
print(lettres[:2])    # du début jusqu'à la position 2 (exclue) : ["a", "b"]
print(lettres[3:])    # de la position 3 jusqu'à la fin : ["d", "e"]
print(lettres[::-1])  # astuce très classique : un "pas" de -1 parcourt la liste à l'envers
```

## 6. Contrôle de flux : décider et répéter

**Une précision avant de commencer** : en Python, l'**indentation** (les espaces au début d'une ligne) n'est pas
une question de style — c'est la syntaxe elle-même. Là où Java utilise des accolades `{ }` pour délimiter un
bloc d'instructions, Python utilise uniquement l'indentation. Une ligne mal indentée est une vraie erreur de
syntaxe, pas juste un code mal présenté.

### 6.1 `if` / `elif` / `else` — décider selon une condition

```python
note = 14

if note >= 16:
    mention = "très bien"
elif note >= 14:        # "elif" = contraction de "else if", vérifié seulement si le if précédent était faux
    mention = "bien"
else:
    mention = "passable"

print(mention)
```

### 6.2 La boucle `for` — répéter une action pour chaque élément d'une collection

En Java, tu écrirais `for (int i = 0; i < 5; i++)`. En Python, l'idée est différente : on ne manipule pas un
compteur qu'on incrémente soi-même, on dit directement "pour chaque élément de telle collection, fais ceci". On
appelle une collection qu'on peut parcourir ainsi un **itérable**.

```python
# range(5) crée une séquence de nombres 0, 1, 2, 3, 4 (5 exclu) — l'équivalent du compteur i de Java
for i in range(5):
    print(i, end=" ")
print()

# Mais on peut aussi itérer directement sur n'importe quel itérable, sans passer par un compteur :
for lettre in "abc":
    print(lettre, end=" ")
print()

for fruit in ["pomme", "banane", "cerise"]:
    print(fruit, end=" ")
print()
```

### 6.3 La boucle `while` — répéter tant qu'une condition reste vraie

Identique dans l'esprit à Java : la condition est vérifiée avant chaque tour de boucle, et la boucle continue
tant qu'elle reste vraie.

```python
compteur = 0
while compteur < 3:
    print("tour", compteur)
    compteur = compteur + 1   # sans cette ligne, la condition resterait vraie pour toujours : boucle infinie
```

### 6.4 La list comprehension — construire une liste en une seule ligne lisible

C'est une façon d'écrire, de manière compacte, exactement le motif "je parcours une collection, et je construis
une nouvelle liste à partir de ça". Regardons d'abord la version explicite, que tu connais déjà, puis la version
compacte équivalente, pour bien voir que c'est rigoureusement la même chose écrite différemment.

```python
# Version explicite, avec une boucle classique :
carres_v1 = []
for n in range(10):
    carres_v1.append(n ** 2)
print(carres_v1)

# Version "list comprehension", strictement équivalente, en une ligne :
carres_v2 = [n ** 2 for n in range(10)]
print(carres_v2)

assert carres_v1 == carres_v2  # on vérifie que les deux versions donnent vraiment le même résultat
```

On peut aussi ajouter un filtre, avec `if`, pour ne garder que certains éléments :

```python
# Version explicite :
carres_pairs_v1 = []
for n in range(10):
    if n % 2 == 0:
        carres_pairs_v1.append(n ** 2)

# Version compacte équivalente :
carres_pairs_v2 = [n ** 2 for n in range(10) if n % 2 == 0]

assert carres_pairs_v1 == carres_pairs_v2
print(carres_pairs_v2)
```

Tu croiseras cette écriture extrêmement souvent dans du code Python professionnel — c'est devenu la façon
"normale" d'écrire ce motif précis, pas juste une astuce de style.

## 7. Les fonctions

**Vocabulaire à fixer d'abord**, parce qu'on confond souvent les deux mots : un **paramètre** est le nom utilisé
*dans la définition* de la fonction (une sorte de variable temporaire qui n'existe que pendant l'exécution de la
fonction) ; un **argument** est la valeur réelle qu'on donne *au moment d'appeler* la fonction. Dans
`def saluer(nom):`, `nom` est le paramètre. Dans `saluer("Youssef")`, `"Youssef"` est l'argument.

```python
def saluer(nom):
    return f"Bonjour {nom}"

print(saluer("Youssef"))
```

**Une confusion fréquente à éviter dès le départ** : `return` et `print` n'ont rien à voir. `print` affiche
quelque chose à l'écran, pour que *toi*, humain, le voies — mais ça ne donne rien à utiliser au reste du
programme. `return` renvoie une valeur que le code appelant peut ensuite stocker dans une variable et réutiliser.
Une fonction sans `return` renvoie silencieusement `None` (vu en section 1).

```python
def qui_affiche(nom):
    print(f"Bonjour {nom}")   # affiche, mais ne renvoie rien d'utilisable ensuite

def qui_renvoie(nom):
    return f"Bonjour {nom}"   # renvoie une valeur utilisable

resultat_1 = qui_affiche("Youssef")   # affiche "Bonjour Youssef" à l'écran pendant l'appel
resultat_2 = qui_renvoie("Youssef")   # n'affiche RIEN à l'écran pendant l'appel

print("resultat_1 vaut :", resultat_1)  # None : qui_affiche n'a rien renvoyé
print("resultat_2 vaut :", resultat_2)  # la vraie chaîne, parce que qui_renvoie l'a renvoyée avec return
```

**Les valeurs par défaut.** Pas de surcharge de méthodes en Python (plusieurs fonctions avec le même nom mais des
paramètres différents, comme en Java). À la place, un paramètre peut avoir une valeur utilisée automatiquement
si on ne la précise pas à l'appel :

```python
def saluer(nom, langue="fr"):   # langue vaudra "fr" si on ne dit rien de plus
    if langue == "fr":
        return f"Bonjour {nom}"
    return f"Hello {nom}"

print(saluer("Youssef"))                 # utilise la valeur par défaut "fr"
print(saluer("Youssef", langue="en"))    # on précise explicitement la valeur, par son nom : plus lisible
```

**`*args` et `**kwargs`** : deux notations spéciales pour accepter un nombre *variable* d'arguments, qu'on ne
connaît pas forcément à l'avance. Tu les verras énormément dans les bibliothèques professionnelles (pandas,
scikit-learn, vus en Brique 4).

```python
def somme(*nombres):   # 'nombres' devient un tuple contenant tous les arguments donnés, quel que soit leur nombre
    total = 0
    for n in nombres:
        total = total + n
    return total

print(somme(1, 2, 3, 4))   # fonctionne avec 4 arguments...
print(somme(10, 20))       # ... ou seulement 2, sans rien changer à la définition de la fonction

def decrire(**infos):      # 'infos' devient un dictionnaire {nom_de_l_argument: valeur}
    for cle, valeur in infos.items():
        print(f"{cle} : {valeur}")

decrire(nom="Youssef", annee="3A", ecole="Polytechnique")
```

**`lambda`** : une façon d'écrire une toute petite fonction, sans lui donner de nom avec `def`, directement là où
on en a besoin — typiquement comme argument d'une autre fonction. Une fonction `lambda` ne peut contenir qu'une
seule expression (pas de plusieurs lignes, pas de `if`/`else` complet).

```python
personnes = [("Alice", 30), ("Bob", 25), ("Chloé", 35)]

# sorted() a besoin de savoir SUR QUOI trier chaque élément : l'argument 'key' attend une fonction
# qui, pour un élément donné, renvoie la valeur à utiliser pour comparer. Ici, p[1] est l'âge.
personnes_triees = sorted(personnes, key=lambda p: p[1])
print(personnes_triees)

# Version équivalente avec une fonction nommée classique, pour bien voir que lambda ne fait rien de magique :
def obtenir_age(p):
    return p[1]

personnes_triees_v2 = sorted(personnes, key=obtenir_age)
assert personnes_triees == personnes_triees_v2
```

## 8. Les chaînes de caractères

**Les f-strings** (`f"..."`) sont la façon moderne, recommandée, d'insérer des valeurs dans un texte — bien plus
lisible que de tout concatener avec `+`.

```python
nom = "Youssef"
age = 22

print(f"{nom} a {age} ans")
print(f"Dans 5 ans, {nom} aura {age + 5} ans")  # on peut écrire une vraie expression entre les accolades
```

Quelques méthodes très courantes sur les chaînes, à connaître par cœur tellement elles reviennent souvent :

```python
phrase = "  Bonjour le Monde  "

print(repr(phrase.strip()))         # strip() retire les espaces inutiles au début et à la fin
print(phrase.strip().lower())       # lower() met tout en minuscules
print(phrase.strip().upper())       # upper() met tout en majuscules
print(phrase.strip().split(" "))    # split(" ") découpe le texte en une liste de morceaux, ici autour des espaces
```

## 9. La gestion des erreurs

**Pourquoi une erreur n'est pas forcément un drame.** Quand Python rencontre une situation impossible à gérer
normalement (par exemple, une division par zéro), il **arrête immédiatement** le programme et affiche un message
décrivant le problème. C'est volontaire : mieux vaut un arrêt net et visible qu'un programme qui continue de
tourner avec un résultat silencieusement faux.

Si tu veux *anticiper* une situation problématique et réagir toi-même plutôt que de laisser le programme
s'arrêter, tu utilises `try` / `except` — l'équivalent du `try`/`catch` de Java, avec une syntaxe un peu
différente :

```python
def diviser(a, b):
    try:
        return a / b
    except ZeroDivisionError:   # ce bloc ne s'exécute QUE si une division par zéro se produit dans le try
        print("Division par zéro, impossible de calculer un résultat.")
        return None

print(diviser(10, 2))   # cas normal : pas d'erreur, le except ne s'exécute pas
print(diviser(10, 0))   # cas problématique : le except intercepte l'erreur au lieu de laisser le programme planter
```

`ZeroDivisionError` est un exemple de **classe d'exception** : Python a plusieurs types d'erreurs différents
selon le problème rencontré (`TypeError` qu'on a vu en Brique 0, `ZeroDivisionError` ici, `ValueError`,
`KeyError` pour une clé absente d'un dictionnaire, etc.). Préciser la classe après `except` permet de ne réagir
qu'à ce type d'erreur précis, et de laisser les autres types d'erreurs arrêter le programme normalement (ce qui
est voulu : on ne veut pas masquer silencieusement des erreurs qu'on n'a pas anticipées).

## 10. Les fichiers

Pour lire ou écrire un fichier, on utilise `open()`. Le mot-clé `with` garantit que le fichier sera bien fermé
automatiquement à la fin, même si une erreur survient pendant qu'on le manipule — l'équivalent du
try-with-resources en Java.

```python
with open("exemple.txt", "w", encoding="utf-8") as f:  # "w" = write, mode écriture (écrase le fichier s'il existe)
    f.write("Bonjour depuis Python\n")

with open("exemple.txt", "r", encoding="utf-8") as f:  # "r" = read, mode lecture
    contenu = f.read()

print(contenu)
```
