### Exercice 1 — Température qui ne descend jamais sous le zéro absolu
Écris une classe `Temperature` dont `__init__` reçoit `celsius` et le stocke via la property (pas directement
dans `_celsius` — passe par `self.celsius = celsius` pour que la vérification s'applique dès la création).
Ajoute une `@property celsius` et son `@celsius.setter`, qui lève `ValueError` si la valeur est inférieure à
`-273.15` (le zéro absolu, en dessous rien n'existe physiquement).

```python
class Temperature:
    def __init__(self, celsius):
        raise NotImplementedError

    @property
    def celsius(self):
        raise NotImplementedError

    @celsius.setter
    def celsius(self, valeur):
        raise NotImplementedError
```

```python
t = Temperature(20)
assert t.celsius == 20

t.celsius = -50
assert t.celsius == -50

try:
    t.celsius = -300
    assert False, "aurait du lever ValueError"
except ValueError:
    pass

try:
    Temperature(-1000)   # la vérification doit s'appliquer dès la création, pas seulement après
    assert False, "aurait du lever ValueError dès __init__"
except ValueError:
    pass

print("OK")
```

### Exercice 2 — Produit avec prix non négatif
Écris une classe `Produit` dont `__init__` reçoit `nom` et `prix`. `prix` est une `@property` dont le setter
lève `ValueError` si la valeur est négative.

```python
class Produit:
    def __init__(self, nom, prix):
        raise NotImplementedError

    @property
    def prix(self):
        raise NotImplementedError

    @prix.setter
    def prix(self, valeur):
        raise NotImplementedError
```

```python
p = Produit("Stylo", 2.5)
assert p.nom == "Stylo"
assert p.prix == 2.5

p.prix = 3.0
assert p.prix == 3.0

try:
    p.prix = -1
    assert False, "aurait du lever ValueError"
except ValueError:
    pass
print("OK")
```

### Exercice 3 — Une property en lecture seule, calculée
Écris une classe `Cercle` dont `__init__` reçoit `rayon` (une `@property` dont le setter refuse les valeurs
inférieures ou égales à 0 — un cercle de rayon nul ou négatif n'a pas de sens). Ajoute une deuxième `@property`,
`aire`, qui **n'a pas de setter** : elle se contente de calculer et renvoyer $\pi r^2$ à partir de `self.rayon`,
à chaque lecture.

*Indice : `import math` puis `math.pi`.*

```python
import math

class Cercle:
    def __init__(self, rayon):
        raise NotImplementedError

    @property
    def rayon(self):
        raise NotImplementedError

    @rayon.setter
    def rayon(self, valeur):
        raise NotImplementedError

    @property
    def aire(self):
        raise NotImplementedError
```

```python
c = Cercle(2)
assert c.rayon == 2
assert abs(c.aire - math.pi * 4) < 1e-9

c.rayon = 3   # changer le rayon doit automatiquement changer l'aire recalculée, sans rien faire de plus
assert abs(c.aire - math.pi * 9) < 1e-9

try:
    c.aire = 100   # aire n'a pas de setter : une tentative d'écriture doit échouer
    assert False, "aurait du lever AttributeError"
except AttributeError:
    pass

try:
    Cercle(-5)
    assert False, "aurait du lever ValueError"
except ValueError:
    pass
print("OK")
```
