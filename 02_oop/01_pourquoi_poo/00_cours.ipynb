# Brique 2.1 — Pourquoi la POO ? Le problème qu'elle résout

On commence cette brique volontairement lentement. Pas de méthodes magiques, pas d'héritage, pas de mot
compliqué aujourd'hui — seulement : **pourquoi la programmation orientée objet existe**, et **la toute première
brique de syntaxe**, construite depuis un exemple concret, pas depuis une définition abstraite.

Le but de ce notebook précis : qu'à la fin, tu puisses dire "une classe sert à ça, concrètement, parce que
j'ai vu de mes yeux le problème qu'elle résout" — pas "une classe c'est un plan pour créer des objets" appris
par cœur sans l'avoir vécu.

## 1. Partons d'un problème concret, sans aucune classe

Imagine que tu dois gérer les informations de plusieurs chiens dans un refuge : leur nom, leur âge, leur race.
Avec ce que tu connais déjà (Brique 1), la façon la plus naturelle de commencer, c'est avec des variables
simples :

```python
chien1_nom = "Rex"
chien1_age = 3
chien1_race = "Labrador"

chien2_nom = "Milo"
chien2_age = 5
chien2_race = "Berger"

print(f"{chien1_nom}, {chien1_age} ans, {chien1_race}")
print(f"{chien2_nom}, {chien2_age} ans, {chien2_race}")
```

Ça marche, pour deux chiens. Maintenant, imagine que le refuge en accueille 50, et qu'on te demande d'écrire une
fonction qui fait vieillir tous les chiens d'un an d'un coup, ou une fonction qui affiche la fiche complète d'un
chien donné. Avec cette approche, il faudrait soit écrire 50 jeux de variables différentes (`chien1_nom`,
`chien2_nom`, ..., `chien50_nom`), soit écrire une fonction avec un nombre de paramètres qui explose. Et rien
n'empêche, par erreur, d'appeler une fonction avec `chien2_nom` mais `chien1_age` — Python ne saura jamais que
ces deux valeurs étaient censées "aller ensemble", puisque rien ne les relie entre elles dans le code, à part le
nom que tu leur as donné par convention.

## 2. Une première amélioration : regrouper avec un dictionnaire

Tu connais déjà le dictionnaire (Brique 1, section 4.3). On peut s'en servir pour regrouper les informations
d'un même chien ensemble, au lieu d'avoir des variables séparées :

```python
chien1 = {"nom": "Rex", "age": 3, "race": "Labrador"}
chien2 = {"nom": "Milo", "age": 5, "race": "Berger"}

chiens = [chien1, chien2]  # maintenant on peut les mettre dans une liste, et en avoir 50 facilement

for chien in chiens:
    print(f"{chien['nom']}, {chien['age']} ans, {chien['race']}")
```

C'est déjà beaucoup mieux : les trois informations d'un même chien voyagent ensemble, dans une seule variable.
On peut maintenant facilement avoir une liste de 50 chiens, et une fonction "faire vieillir tous les chiens"
devient simple à écrire.

**Mais il reste deux problèmes, plus subtils.**

**Premier problème : rien ne garantit la structure.** Rien n'empêche quelqu'un (ou toi-même, trois mois plus
tard) d'écrire `{"nom": "Rex", "age": 3}` en oubliant `"race"`, ou `{"nom": "Rex", "Age": 3, "race": "Labrador"}`
avec une majuscule à `Age` par erreur. Le dictionnaire ne décrit jamais "à quoi doit ressembler un chien" — il se
contente d'accepter n'importe quelles clés qu'on lui donne, bonnes ou fausses, sans jamais rien vérifier.

**Deuxième problème, plus important : les données et les actions qu'on fait dessus sont séparées.** Si tu veux
écrire une fonction "faire vieillir ce chien d'un an", elle vit quelque part ailleurs dans ton code, complètement
déconnectée du dictionnaire lui-même :

```python
def vieillir(chien):
    chien["age"] = chien["age"] + 1

vieillir(chien1)
print(chien1)
```

Cette fonction `vieillir` n'a aucun lien officiel avec les chiens : rien n'empêche de l'appeler par erreur sur
n'importe quel autre dictionnaire qui n'a rien à voir (`vieillir({"titre": "un livre"})` planterait seulement au
moment de l'exécution, pas avant — on retrouve ici l'idée du typage dynamique vue en Brique 0). Et si demain tu
as aussi des `chat`, des `oiseau`, chacun avec ses propres fonctions (`vieillir_chat`, `vieillir_oiseau`...), tout
ce petit monde de données et de fonctions traîne séparément, sans structure qui dit clairement "ces fonctions-là
vont avec ce type de données-là".

## 3. L'idée centrale de la POO : regrouper les données ET les actions, au même endroit

Voici l'idée que toute la programmation orientée objet construit à partir de là : au lieu de séparer "les
données" (le dictionnaire) et "les actions sur ces données" (les fonctions à côté), **on définit, à un seul
endroit, un modèle précis qui dit à la fois : quelles informations un chien possède, ET quelles actions on peut
faire sur un chien.** Ce modèle s'appelle une **classe**.

**Une analogie pour fixer la différence entre deux mots qu'on va utiliser tout le temps à partir de maintenant :
classe et objet (ou "instance").**

Pense à un plan d'architecte pour une maison. Le plan décrit, en général, ce qu'une maison de ce modèle précis
aura : un nombre de pièces, une surface, des fenêtres à tel endroit. Le plan lui-même n'est pas une maison —
on ne peut pas y habiter. Mais à partir de ce même plan, on peut construire autant de vraies maisons qu'on veut,
chacune à une adresse différente, avec sa propre couleur de façade, son propre numéro de rue — chacune étant un
exemplaire concret et distinct, construit en suivant ce plan.

- La **classe**, c'est le plan : elle décrit la structure (quelles informations) et le comportement (quelles
  actions possibles), pour tous les chiens qui suivront ce modèle.
- Un **objet** (on dit aussi une **instance** de la classe) c'est une vraie maison construite à partir de ce
  plan : un chien bien réel et concret, avec son propre nom, son propre âge — mais qui respecte la structure
  décrite par la classe `Chien`.

On peut créer autant d'objets qu'on veut à partir d'une seule classe, exactement comme on peut construire
plusieurs maisons différentes à partir d'un seul plan.

## 4. Construire notre première classe, étape par étape

On va construire la classe `Chien` par petites touches successives, en observant à chaque étape ce qui change,
plutôt que de l'écrire d'un bloc.

**Étape 1 : déclarer la classe, vide.**

```python
class Chien:
    pass  # 'pass' veut dire "ne rien faire" : on a besoin d'au moins une instruction, même vide, pour que ce soit valide

mon_chien = Chien()  # on crée un objet (une instance) à partir de la classe, avec des parenthèses, comme un appel de fonction
print(mon_chien)
print(type(mon_chien))
```

Déjà, deux choses à observer : `type(mon_chien)` confirme bien que `mon_chien` est de type `Chien` — exactement
comme `type(42)` renvoyait `int` en Brique 1. Et `print(mon_chien)` affiche quelque chose comme
`<__main__.Chien object at 0x...>` : un affichage peu lisible par défaut, qu'on saura améliorer plus tard (dans
le notebook sur les méthodes spéciales — pas aujourd'hui, on garde les choses simples).

Pour l'instant, cette classe ne sert à rien : elle ne décrit encore aucune information, aucun comportement.

**Étape 2 : donner à chaque chien ses propres informations, avec `__init__` et `self`.**

On veut que chaque objet `Chien` créé ait, dès sa naissance, un nom, un âge, une race. Pour ça, on écrit une
méthode spéciale appelée `__init__` (construite avec deux underscores de chaque côté — "init" pour
"initialisation"). Python l'appelle **automatiquement**, une seule fois, exactement au moment où un nouvel objet
est créé.

Mais avant de l'écrire, il faut construire un mot qui va apparaître partout à partir de maintenant : **`self`**.

### Comprendre `self`, lentement, parce que c'est LE point qui bloque le plus souvent

**Le problème que `self` résout.** Une classe décrit le modèle commun à *tous* les chiens (le plan). Mais quand
on écrit une méthode comme "vieillir d'un an", il faut bien que cette méthode sache **sur quel chien précis**
elle doit agir — sur `mon_chien`, ou sur un autre objet `Chien` créé à partir de la même classe ? La classe elle-
même ne le sait pas à l'avance : elle décrit la règle générale ("vieillir veut dire ajouter 1 à l'âge"), mais
cette règle doit s'appliquer à un objet précis au moment où on l'utilise réellement.

**La solution de Python : donner explicitement, en premier paramètre de chaque méthode, une référence vers
l'objet concerné.** Par convention (pas une obligation stricte du langage, mais une règle suivie par absolument
tout le monde), ce premier paramètre s'appelle `self` — littéralement "soi-même", "cet objet précis, celui sur
lequel la méthode est en train d'agir".

**Une précision importante si tu as déjà vu du Java ou cherché des explications en ligne** : en Java, l'objet
courant est accessible via le mot `this`, mais tu n'as jamais besoin de le déclarer explicitement comme
paramètre d'une méthode — Java le fournit automatiquement en coulisses. **Python fait un choix différent et
explicite** : il n'y a rien de magique ni caché, `self` est un paramètre normal, comme n'importe quel autre, que
tu dois écrire toi-même en premier dans chaque méthode. Rien n'oblige à l'appeler `self` (c'est juste une
convention universelle, ne t'en écarte jamais), mais sa position en premier paramètre, elle, est obligatoire.

**Vérifions-le concrètement avant d'aller plus loin**, avec un exemple qui va démonter la mécanique derrière
`self`, pour que ce ne soit plus du tout mystérieux :

```python
class Chien:
    def aboyer(self):
        print(f"{self} aboie !")

mon_chien = Chien()

# La façon normale d'appeler une méthode : avec un point, sur l'objet.
mon_chien.aboyer()

# Mais ce que Python fait VRAIMENT en coulisses, c'est ceci, de façon strictement équivalente :
Chien.aboyer(mon_chien)  # on appelle la méthode "depuis la classe", en lui donnant explicitement l'objet concerné

# Les deux lignes ci-dessus font EXACTEMENT la même chose. mon_chien.aboyer() n'est qu'un raccourci
# d'écriture pour Chien.aboyer(mon_chien). Le "self" dans la définition de aboyer, c'est précisément
# ce qui reçoit mon_chien quand on utilise la version avec le point.
```

Garde cette équivalence en tête chaque fois que `self` te semblera flou : `objet.methode(argument)` est toujours,
sous le capot, strictement la même chose que `Classe.methode(objet, argument)`. `self`, ce n'est rien de plus que
"l'objet avec lequel on a appelé cette méthode", automatiquement glissé en premier.

### Maintenant, écrivons vraiment `__init__`

`__init__` est une méthode comme une autre (donc elle aussi reçoit `self` en premier paramètre), mais Python
l'appelle tout seul, automatiquement, dès qu'on écrit `Chien(...)`. Son rôle : préparer les informations de
*cet* objet précis, au moment de sa création.

```python
class Chien:
    def __init__(self, nom, age, race):
        # self.nom = nom : on crée un attribut 'nom', propre à CET objet self, et on lui donne
        # la valeur reçue en paramètre. 'self.nom' et 'nom' sont deux choses différentes :
        # 'nom' n'existe que temporairement, pendant l'exécution de __init__ ;
        # 'self.nom' est enregistré durablement SUR l'objet, accessible même après que __init__ soit terminé.
        self.nom = nom
        self.age = age
        self.race = race

chien1 = Chien("Rex", 3, "Labrador")   # Python appelle automatiquement __init__(self=chien1, nom="Rex", age=3, race="Labrador")
chien2 = Chien("Milo", 5, "Berger")

print(chien1.nom, chien1.age, chien1.race)  # on accède à un attribut avec un point, comme pour appeler une méthode
print(chien2.nom, chien2.age, chien2.race)
```

**Vérifions bien que `chien1` et `chien2` sont deux objets séparés, chacun avec ses propres valeurs**, et pas une
seule case mémoire partagée par erreur :

```python
chien1.age = chien1.age + 1  # on modifie SEULEMENT chien1
print("chien1 :", chien1.nom, chien1.age)  # 4
print("chien2 :", chien2.nom, chien2.age)  # toujours 5, chien2 n'a pas bougé
```

**Vocabulaire à fixer** : `nom`, `age`, `race` sont appelés des **attributs** de la classe `Chien` (des
informations rattachées à chaque objet). `__init__` et `aboyer` sont appelées des **méthodes** (des actions
qu'on peut faire sur un objet). Les attributs, ce sont les données ; les méthodes, ce sont les comportements —
exactement les deux choses qu'on voulait regrouper ensemble en section 3.

**Étape 3 : ajouter une vraie méthode qui fait quelque chose d'utile.**

On reprend l'exemple du début (faire vieillir un chien), mais cette fois **la fonction vit à l'intérieur de la
classe**, directement attachée à ce qu'elle manipule :

```python
class Chien:
    def __init__(self, nom, age, race):
        self.nom = nom
        self.age = age
        self.race = race

    def vieillir(self):
        self.age = self.age + 1

    def se_presenter(self):
        return f"Je m'appelle {self.nom}, j'ai {self.age} ans, je suis un {self.race}"

chien1 = Chien("Rex", 3, "Labrador")
print(chien1.se_presenter())

chien1.vieillir()
print(chien1.se_presenter())
```

Compare avec la section 2 : `vieillir` n'est plus une fonction isolée qui pourrait recevoir n'importe quel
dictionnaire par erreur. Elle est maintenant une partie de la classe `Chien` elle-même, elle n'existe que pour
les objets `Chien`, et `self` garantit qu'elle agit toujours sur l'objet précis avec lequel on l'appelle.
